```python
import pandas as pd
train = pd.read_csv('https://raw.githubusercontent.com/milena-seismo/DSB-Unit-02-New/refs/heads/main/Labs/lab-eda-titanic/titanic-train.csv')
```

```python
train_new = train.copy()
```

```python
train_new.head()
```

```python
#There are two columns that pertain to how many family members are on the boat for a given person. Create a new column called FamilyCount which will be the sum of those two columns.
train_family = train_new.copy()
train_family['FamilyCount'] = train_family['SibSp'] + train_family['Parch']
```

```python
train_family["FamilyCount"]
```

```python
train_family[["Name","FamilyCount"]].head(20)
```

```python
# Reverends have a special title in their name. Create a column called IsReverend: 1 if they're a preacher, 0 if they're not.
train_family['IsReverend'] = 0

for i in train_family.index:
    if 'Rev.' in train_family.loc[i, 'Name']:
        train_family.loc[i, 'IsReverend'] = 1
```

```python
train_family['IsReverend']
```

```python
train_family.groupby('IsReverend')['Name'].unique()
```

```python
train_family.loc[train_family['IsReverend'] == 1, 'Name']
```

```python
# Create 3 columns: Embarked_C, Embarked_Q and Embarked_S. These columns will have 1's and 0's that correspond to the C, Q and S values in the Embarked column

embarked = pd.get_dummies(
    train_family['Embarked'],
    prefix='Embarked',
    dtype=int
)
```

```python
# The same thing for Sex

sex = pd.get_dummies(
    train_family['Sex'],
    prefix='Sex',
    dtype=int
)
```

```python
# What was the survival rate for each Pclass

train_survival = train.groupby('Pclass')['Survived'].mean() * 100

print(train_survival)
```

```python
# How many reverends survive?
train_family = train.copy()

train_family['IsReverend'] = 0

for i in train_family.index:
    if 'Rev.' in train_family.loc[i, 'Name']:
        train_family.loc[i, 'IsReverend'] = 1

```

```python
train_family.groupby('IsReverend')['Name'].unique()
```

```python
# What is the survival rate for cabins marked ¯\(ツ)/¯

train_family.columns
```

```python
train_family.isnull().sum()
```

```python
train_family['Cabin'] = train_family['Cabin'].fillna(r'¯\_(ツ)_/¯')
```

```python
train_family['Cabin'].value_counts()
```

```python
# What is the survival rate for cabins marked ¯\(ツ)/¯

survival_rate = train_family.loc[train_family['Cabin'] == r'¯\_(ツ)_/¯', 'Survived'].mean() * 100

print(f'Survival rate: {survival_rate:.1f}%')


```

```python
# What is the survival rate for people whose Age is empty?


survival_rate = train_family.loc[
    train_family['Age'].isnull(),
    'Survived'
].mean() * 100

print(f'Survival rate: {survival_rate:.1f}%')
```

```python
train_family['Age'].head(22)
```

```python
# What is the survival rate for each port of embarkation

train_survival = train_family.groupby('Embarked')['Survived'].mean() * 100

print(train_survival)
```

```python
# What is the survival rate for children (under 12) in each Pclass?

train_survival = train_family.loc[train_family['Age'] < 12].groupby('Pclass')['Survived'].mean() * 100

print(train_survival)
```

```python
# Did the captain of the ship survive? Is he on the list? Ai suggested to find the name of captain. So captain of Titanic was called Edward Smit.


for name in train_family['Name']:
    if 'Smith' in name:
        print(name)


# He wos not on the passenger list.
#From historical data we know that he did not survive
```

```python
# Of all the people that died, who had the most expensive ticket? How much did it cost?

died = train_family.loc[train_family['Survived'] == 0]

expensive = died['Fare'].max()

print(died.loc[died['Fare'] == expensive, ['Name', 'Fare']])
```

```python
# Does having family on the boat help or hurt your chances of survival?


train_family['FamilyCount'] = train_family['SibSp'] + train_family['Parch']
```

```python
train_family['FamilyCount'].value_counts()
```

```python
alone = train_family.loc[train_family['FamilyCount'] == 0]
family = train_family.loc[train_family['FamilyCount'] > 0]

print(f'Alone: {alone["Survived"].mean() * 100:.1f}%')
print(f'With family: {family["Survived"].mean() * 100:.1f}%')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.barplot(data=train_family, x='Sex', y='Survived')

plt.title('Survival Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('Survival Rate')
plt.show()
```

```python
sns.barplot(data=train_family, x='Pclass', y='Survived')

plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.show()
```

```python
sns.barplot(data=train_family, x='Pclass', y='Survived', hue='Sex')

plt.title('Survival Rate by Class and Gender')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.show()
```
